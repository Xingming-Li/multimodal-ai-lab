```python
import torch
from PIL import Image
from IPython.display import display
import torchvision.transforms as transforms
```

```python
# Read and show a PIL image
image = Image.open('sweden.jpg')
display(image)
```

```python
# Define a transform to convert PIL image to a Torch tensor
transform = transforms.Compose([transforms.PILToTensor()])
# transform = transforms.PILToTensor()

# Convert the PIL image to Torch tensor
img_tensor = transform(image)

# Print the shape and the converted Torch tensor
print(img_tensor.shape)
print(img_tensor)
```

```python
# Transform back to image
transform_back = transforms.ToPILImage()
image_back = transform_back(img_tensor)
display(image_back)
```

```python
# Clone the tensor
cloned = img_tensor.clone()

# Modify the cloned tensor
modified = torch.where(cloned > 50, cloned - 50, cloned)

# Transform back to see differences
image_mod = transform_back(modified)
display(image_mod)
```
